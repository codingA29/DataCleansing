```python
file_path = "/content/drive/MyDrive/Dataset_Pasien_Latihan_Data_Cleansing.xlsx"
```

```python
import pandas as pd

df = pd.read_excel(
    file_path,
    sheet_name="Dataset Mentah"
)

display(df)
```

```python
# Melihat ukuran dataset
print("Jumlah baris dan kolom:", df.shape)

# Melihat informasi dataset
print("\nInformasi Dataset:")
df.info()
```

```python
# Mengecek jumlah missing value pada setiap kolom
print("Jumlah Missing Value:")
print(df.isnull().sum())
```

```python
# Menampilkan data yang memiliki nilai kosong
data_missing = df[df.isnull().any(axis=1)]

print("Data yang memiliki Missing Value:")
display(data_missing)
```

```python
# Mengisi missing value pada Umur dengan nilai median
median_umur = df["Umur"].median()
df["Umur"] = df["Umur"].fillna(median_umur)

# Mengisi missing value pada Diagnosis dengan nilai modus
modus_diagnosis = df["Diagnosis"].mode()[0]
df["Diagnosis"] = df["Diagnosis"].fillna(modus_diagnosis)

print("Median Umur:", median_umur)
print("Modus Diagnosis:", modus_diagnosis)

print("\nJumlah Missing Value Setelah Cleansing:")
print(df.isnull().sum())
```

```python
# Menampilkan kembali data PS012 dan PS023
display(
    df[df["ID Pasien"].isin(["PS012", "PS023"])]
)
```

```python
print("Total Missing Value:", df.isnull().sum().sum())
```

```python
# Mengecek jumlah data duplikat
jumlah_duplikat = df.duplicated().sum()

print("Jumlah Data Duplikat:", jumlah_duplikat)
```

```python
# Menampilkan semua data yang terduplikasi
data_duplikat = df[df.duplicated(keep=False)]

print("Data yang Terduplikasi:")
display(data_duplikat)
```

```python
# Menghapus data duplikat
df = df.drop_duplicates()

print("Data duplikat berhasil dihapus")
```

```python
# Mengecek kembali data duplikat
print("Jumlah Duplikat Setelah Cleansing:", df.duplicated().sum())

print("Jumlah Data Setelah Duplikat Dihapus:", len(df))
```

```python
# Mengecek nilai unik sebelum standardisasi

print("Data Unik Diagnosis:")
print(df["Diagnosis"].unique())

print("\nData Unik Status Rawat:")
print(df["Status Rawat"].unique())
```

```python
# Membersihkan spasi berlebih pada kolom teks

kolom_teks = [
    "ID Pasien",
    "Nama Pasien",
    "Diagnosis",
    "Status Rawat"
]

for kolom in kolom_teks:
    df[kolom] = df[kolom].str.strip()

print("Spasi berlebih berhasil dibersihkan.")
```

```python
# Menyeragamkan format penulisan data

df["ID Pasien"] = df["ID Pasien"].str.upper()
df["Nama Pasien"] = df["Nama Pasien"].str.title()
df["Diagnosis"] = df["Diagnosis"].str.title()
df["Status Rawat"] = df["Status Rawat"].str.title()

print("Standardisasi format teks berhasil dilakukan.")
```

```python
print("Diagnosis Setelah Standardisasi:")
print(df["Diagnosis"].unique())

print("\nStatus Rawat Setelah Standardisasi:")
print(df["Status Rawat"].unique())
```

```python
# Memperbaiki kesalahan penulisan diagnosis

df["Diagnosis"] = df["Diagnosis"].replace({
    "Jantng": "Jantung"
})

print("Diagnosis Setelah Perbaikan Typo:")
print(df["Diagnosis"].unique())
```

```python
# Menentukan kategori diagnosis yang valid

diagnosis_valid = [
    "Penyakit Dalam",
    "Jantung",
    "Bedah/Ortopedi",
    "Infeksi",
    "Saraf"
]

diagnosis_tidak_valid = df[
    ~df["Diagnosis"].isin(diagnosis_valid)
]

print("Jumlah Diagnosis Tidak Valid:",
      len(diagnosis_tidak_valid))

display(diagnosis_tidak_valid)
```

```python
# Menentukan kategori status rawat yang valid

status_valid = [
    "Rawat Jalan",
    "Rawat Inap"
]

status_tidak_valid = df[
    ~df["Status Rawat"].isin(status_valid)
]

print("Jumlah Status Rawat Tidak Valid:",
      len(status_tidak_valid))

display(status_tidak_valid)
```

```python
# Mengecek umur pasien yang tidak valid

umur_tidak_valid = df[
    (df["Umur"] < 8) |
    (df["Umur"] > 68)
]

print("Data dengan Umur Tidak Valid:")
display(umur_tidak_valid)
```

```python
# ==========================================
# PROSES FINAL DATA CLEANSING
# ==========================================

# 1. Menghapus data dengan umur tidak valid
df = df[
    (df["Umur"] >= 8) &
    (df["Umur"] <= 68)
]

# 2. Reset index
df = df.reset_index(drop=True)

# 3. Menampilkan dataset bersih
print("=== DATASET HASIL DATA CLEANSING ===")
print("Jumlah Data Bersih:", len(df))

display(df)
```

```python
# Menyimpan dataset bersih ke Google Drive

output_path = (
    "/content/drive/MyDrive/"
    "Dataset_Pasien_Hasil_Cleansing.xlsx"
)

df.to_excel(
    output_path,
    index=False
)

print("Dataset bersih berhasil disimpan.")
print("Lokasi file:")
print(output_path)
```

```python
import pandas as pd

file_path = (
    "/content/drive/MyDrive/"
    "Dataset_Pasien_Hasil_Cleansing.xlsx"
)

df = pd.read_excel(file_path)

print("Dataset berhasil dibaca.")
display(df)
```

```python
def kelompok_umur(umur):
    if umur <= 12:
        return "Anak"
    elif umur <= 17:
        return "Remaja"
    elif umur <= 59:
        return "Dewasa"
    else:
        return "Lansia"
```

```python
df["Kelompok Umur"] = df["Umur"].apply(kelompok_umur)

print("Kolom Kelompok Umur berhasil ditambahkan.")
```

```python
print("=== HASIL DATA ENRICHMENT ===")

display(df)
```

```python
print("Jumlah Pasien Berdasarkan Kelompok Umur:")

print(df["Kelompok Umur"].value_counts())
```

```python
output_path = (
    "/content/drive/MyDrive/"
    "Dataset_Pasien_Hasil_Enrichment.xlsx"
)

df.to_excel(
    output_path,
    index=False
)

print("Dataset hasil enrichment berhasil disimpan.")
print("Lokasi file:", output_path)
```
